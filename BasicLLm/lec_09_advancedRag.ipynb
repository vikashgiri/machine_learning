What is GraphRAG?

GraphRAG is an advanced version of RAG that combines:

Vector Search (for semantic similarity)
Knowledge Graph (for entities and relationships)

It helps the system understand not just similar text, but also how things are connected.

User Question
      ↓
┌──────────────────┬──────────────────┐
│  Vector Search   │   Graph Search   │
│  (Meaning)       │  (Relationships) │
└──────────────────┴──────────────────┘
      ↓                    ↓
        Combined Knowledge
               ↓
              LLM
               ↓
           Final Answer

           Why Normal RAG is Not Enough
Normal RAG is good when the answer is in one place.
It becomes weak when:

The answer requires connecting multiple pieces of information
The question needs understanding of relationships
Multi-hop reasoning is required

Example Question:
“Which products were developed by the team that the current CTO used to lead?”
This kind of question is hard for normal vector search but easier for GraphRAG.

Core Idea of GraphRAG

Extract Entities from documents
(People, Companies, Products, Policies, Concepts, etc.)
Extract Relationships between them
(works_at, manages, part_of, related_to, etc.)
Build a Knowledge Graph
During query time:
Find relevant entities/text using vector search
Explore related entities using the graph
Give both textual context + graph connections to the LLM

Simple Example
Documents contain:

“Alice was the head of Team Phoenix from 2018 to 2021.”
“Team Phoenix developed Product X and Product Y.”
“Bob became CTO in 2023. He previously worked with Alice.”

Knowledge Graph:
(Alice) — led → (Team Phoenix)
(Team Phoenix) — developed → (Product X)
(Team Phoenix) — developed → (Product Y)
(Bob) — worked_with → (Alice)
(Bob) — is → (CTO)

Situation                           Use GraphRAG?
Simple document Q&A                     No
Questions about relationships         Yes
Multi-hop reasoning needed            Yes
Large enterprise knowledge base       Yes
Need explainability                   Yes
Small and simple documents           Not needed

Now GraphRAG can answer relationship-based questions much better.

What is Multi-Modal RAG?

Multi-Modal RAG is a type of RAG that can work with multiple types of data, not just text.
It can retrieve and understand:

Text Based Rag can not understand Images like Charts, architecture diagrams, Tables, Charts / Diagrams,Code,(Sometimes audio or video) and also Hard to understand structure and it Completely ignored so Multi-Modal RAG solves this by handling all these content types.

User Question
      ↓
Retrieve relevant content from:
  - Text chunks
  - Images
  - Tables
  - Code blocks
      ↓
Send everything to a Multi-modal LLM
      ↓
Final Answer

How Multi-Modal RAG Works

Parsing

Extract text, tables, images, and code from documents (PDF, PPT, HTML, etc.)

Processing each type

Text → Normal chunking + text embedding
Images → Captioning or image embedding (CLIP, etc.)
Tables → Convert to text/markdown or keep structure
Code → Special code-aware chunking

Storage

Store all of them in a vector database (with metadata about type)

Retrieval

Retrieve the most relevant items across all modalities

Generation

Send the retrieved text + images + tables to a multi-modal LLM (like GPT-4o, Claude, Gemini)

These are two different ways to handle complex questions in RAG systems.

1. Long-Context Approach
 
 Put a large amount of text (or even the full document) directly into the LLM’s context window and let the model find the answer

 User Question + Large Context → LLM → Answer

 Modern LLMs support very long context (128k, 200k, even 1M tokens).
Pros

Simple to implement
Good for single long documents
Less engineering effort
No complex retrieval logic needed

Cons

Expensive (more tokens = higher cost)
Slower
Model can miss information (“lost in the middle”)
Hard to scale across many documents
Weak when information is spread across multiple sources

2. Multi-Hop Retrieval
Idea:
Retrieve information in multiple steps. The system gathers pieces of information one by one and combines them.

Question
   ↓
Retrieve first piece of information
   ↓
Use it to find the next relevant piece
   ↓
Repeat if needed
   ↓
Combine everything → Final Answer

Pros

Better for multi-hop / complex questions
More scalable
Usually cheaper than long context
More controlled and explainable
Works well with query decomposition and GraphRAG

Cons

More complex to build
Higher latency (multiple retrieval rounds)
Intermediate steps can go wrong

Example

“Which products were built by the team that the current CTO previously led?”

Approach          How it works

Long-Context     Put the entire long document(s) into the LLM

Multi-Hop   "1st hop: Find who is the CTO 2nd hop: Find the team he previously led 3rd hop: Find    


products of that team"

When to Use Which?:-

Situation                                             Better Choice
One long document                                   Long-Context
Information spread across many documents            Multi-Hop Retrieval
Need high scalability                              Multi-Hop Retrieval
Want simple system                                 Long-Context
Complex relationship questions                     Multi-Hop / GraphRAG
Cost-sensitive system                               Multi-Hop Retrieval
Very high accuracy on complex questions           Multi-Hop Retrieval


Current Trend (2026)

Long-context models are getting better and cheaper.
For serious enterprise systems with many documents, Multi-Hop Retrieval is still preferred.
Many production systems use a hybrid approach:
Try simple retrieval first → If not enough, switch to multi-hop.

There is no single winner which is better. It depends on your use case.

What is Adaptive RAG?

Adaptive RAG means the system dynamically decides how to handle each question, instead of using the same retrieval strategy for every query.

User Question
      ↓
Analyze the question
      ↓
Decide the best strategy:
  - No retrieval needed
  - Simple vector search
  - Hybrid search
  - Multi-hop retrieval
  - GraphRAG
  - Long-context
      ↓
Execute the chosen strategy
      ↓
Generate Answer

It adapts based on the type and complexity of the question.

How Adaptive RAG Works

Query Analysis
An LLM or classifier analyzes the question.
Strategy Selection
Decides the best retrieval method.
Execution
Runs only the selected strategy.
Generation
Produces the final answer.

Example
Question 1:
“What is the capital of France?”
→ No retrieval needed (LLM already knows)
Question 2:
“How many casual leaves do employees get?”
→ Simple Vector Search
Question 3:
“Which products were developed by the team that the current CTO previously led?”
→ Multi-Hop Retrieval or GraphRAG



What is Corrective RAG?

Corrective RAG (CRAG) is an advanced RAG technique that checks the quality of retrieved documents and takes corrective actions if the retrieval is poor.

User Question
      ↓
Retrieve documents
      ↓
Evaluate Retrieval Quality
      ↓
┌─────────────────┬────────────────────┬────────────────────┐
│  Good Retrieval │  Ambiguous         │  Poor Retrieval    │
│  → Use docs     │  → Refine / Re-search │ → Use Web Search │
└─────────────────┴────────────────────┴────────────────────┘
      ↓
Generate Final Answer

It tries to correct bad retrieval before generating the answer.

Why Do We Need Corrective RAG?
Normal RAG has a big problem:

If retrieval is bad → Final answer becomes bad
The LLM still tries to answer using poor context → Hallucination risk increases

Corrective RAG adds a quality check step to fix this.

How Corrective RAG Works

Retrieve documents using vector search
Evaluate how relevant the retrieved documents are
Decide one of the following actions:
Correct → Documents are good → Use them
Incorrect → Documents are bad → Trigger alternative strategy (usually web search or refined search)
Ambiguous → Partially useful → Refine the query or combine sources

Generate the final answer

What is Self-RAG?

Self-RAG (Self-Reflective Retrieval-Augmented Generation) is an advanced RAG technique where the model reflects on its own retrieval and generation process.
The system decides:

Whether it needs to retrieve documents or not
Whether the retrieved documents are relevant
Whether the generated answer is good enough
Whether it should retrieve again or refine the answer

User Question
      ↓
Decide: Do I need retrieval?
      ↓
Retrieve documents (if needed)
      ↓
Evaluate retrieved documents (Relevant or not?)
      ↓
Generate Answer
      ↓
Evaluate the Answer (Is it supported? Is it good?)
      ↓
If not good → Retrieve again or refine
      ↓
Final Answer

Key Idea of Self-RAG
Normal RAG always retrieves documents.
Self-RAG makes the model think and control the process itself.
It uses special reflection tokens, such as:

Retrieve / No Retrieve
Relevant / Irrelevant
Supported / Not Supported
Useful / Not Useful


How Self-RAG Works (Step-by-Step)

Retrieval Decision
Model decides if it needs external documents or can answer directly.

Retrieval
If needed, it retrieves documents.

Document Evaluation
Checks if the retrieved documents are relevant.

Generation
Generates an answer using the relevant documents.

Self-Critique
Evaluates if the answer is factually supported by the documents.

Correction (if needed)
If the answer is not good, it can retrieve again or regenerate.

What is Agentic RAG?

Agentic RAG is an advanced form of RAG where an AI Agent controls the retrieval and reasoning process.
Instead of a fixed pipeline, the system acts like an intelligent agent that can:

Decide when to retrieve
Decide what to retrieve
Use multiple tools
Perform multi-step reasoning
Critique and improve its own answers

User Question
      ↓
Agent thinks
      ↓
Decides actions:
  - Search documents
  - Search web
  - Query database
  - Use calculator
  - Call APIs
  - Reflect on results
      ↓
Executes tools step by step
      ↓
Generates Final Answer


How Agentic RAG Works

User asks a question
Agent analyzes the question
Plans the steps needed
Selects tools (document search, web search, etc.)
Executes tools one by one
Observes the results
Reflects — Is the information enough?
If not → takes more actions
Finally generates the answer

Example
Question:
“What is our current casual leave policy, and how does it compare with the industry standard in 2026?”
Agentic RAG might do:

Search internal documents for company leave policy
Search the web for industry standards in 2026
Compare both
Generate a final comparison answer

Normal RAG usually cannot do this well.

What is Incremental Indexing?

Incremental Indexing means updating the vector database only with the changes, instead of re-processing and re-embedding everything from scratch.

Old Way (Full Re-indexing):
Process all documents again → Create all embeddings again → Replace entire index

Incremental Way:
Detect what changed → Process only new/updated/deleted content → Update only those parts

In production, documents keep changing. Full re-indexing is usually not practical.

What is Change Detection?
Change Detection is the process of identifying what has changed in your source documents.
You need to detect:

Change Type,           Description
New                  New documents/chunks added
Updated              Existing documents were modified
Deleted              Documents were removed
Unchanged            No action needed

change detection method is Content Hashing which Create hash (MD5/SHA) of content and compare but Need to store hashes


How Incremental Indexing Works

Scan source documents
Detect changes (New / Updated / Deleted)
For New documents:
Chunk → Embed → Insert into vector DB

For Updated documents:
Delete old chunks of that document
Re-chunk → Re-embed → Insert new chunks

For Deleted documents:
Delete all related chunks from vector DB

Update the tracking records (hashes / timestamps)

Incremental Indexing = Only process what has changed
Change Detection = Identify New, Updated, and Deleted content
Critical for production RAG systems
Saves time, money, and compute
Content hashing is one of the most reliable methods

2. Temporal RAG

Temporal RAG means the system can answer questions with time awareness.
It understands:

What information was valid at a certain time
Current vs historical information
Time-based queries

Examples of Temporal Questions:
- What is the current casual leave policy?
- What was the leave policy in March 2025?
- How has the maternity leave policy changed since 2024?

How Temporal RAG Works

Store documents with time metadata (valid_from, valid_to, version)
User asks a time-related question
System extracts the time intent from the question
Filters documents based on the relevant time range
Performs retrieval only on the valid documents for that time
Generates the answer

Example
Question:
“What was the casual leave policy in 2025?”
System does:

Detects time → Year 2025
Filters documents where valid_from <= 2025 and valid_to >= 2025
Retrieves only from those versions
Answers based on the 2025 policy

Best Practices

Always store valid_from and valid_to
Keep document_id same across versions
Mark the latest version with is_current = true
For updates: Close previous version (valid_to = now) and insert new version
Support both “current” and “as-of-date” queries
Combine with Incremental Indexing for efficiency

Together, they allow your RAG system to correctly handle questions about the past, present, and changes over time.

What is Context Packing?

Context Packing means selecting and arranging the retrieved information in the smartest way so that the most useful content fits inside the LLM’s token limit.

Retrieved Chunks (can be many)
        ↓
Context Packing (select + order + compress)
        ↓
Final Context (fits within token limit)
        ↓
Send to LLM

Even if you retrieve 20 good chunks, you usually cannot send all of them because of token limits.

Main Goals of Context Packing

Stay within the token limit
Keep the most relevant information
Maintain useful order
Reduce noise
Maximize answer quality

Common Context Packing Strategies
1. Top-K Selection (Basic)

Take top 3–5 most relevant chunks
Simple but can miss useful information

2. Relevance Score Filtering

Only keep chunks above a similarity threshold
Removes weak matches

3. Maximal Marginal Relevance (MMR)

Balances Relevance + Diversity
Avoids sending multiple very similar chunks

Best Practice Flow (Production)
1. Retrieve more candidates (e.g. top 20)
2. Rerank them
3. Remove highly similar / redundant chunks (MMR)
4. Apply compression if needed
5. Pack the best chunks within token limit
6. Order them properly (best info not buried in the middle)
7. Send to LLM

Example Token Budget
Assume model context limit = 8,000 tokens

Part              Tokens Allocated
SystemPrompt         400
User Question        100
Retrieved Context     5,500"
Safety buffer         500
Reserved for Answer   1,500"

You must pack context within ~5,500 tokens.

Practical Tips
Always retrieve more than you will finally use (e.g. retrieve 15–30 → pack 4–8)
Use reranking before packing

What is Structured Output?

Structured Output means making the LLM return responses in a fixed, machine-readable format (like JSON, XML, or a specific schema) instead of free-form text.

What is Constrained Generation?

Constrained Generation means controlling the LLM so that it can only generate outputs that follow specific rules or a schema.
You restrict the model’s output space so it doesn’t produce invalid formats.

LLM can only generate valid JSON according to the given schema

What is Multi-Step Reasoning over Retrieved Evidence?

It means the system does not just retrieve documents and generate an answer in one step.
Instead, it:

Retrieves evidence
Thinks about the evidence
May retrieve more information based on what it learned
Chains multiple reasoning steps
Finally arrives at the answer

Question
   ↓
Retrieve Evidence
   ↓
Reason over Evidence
   ↓
Need more information? → Yes → Retrieve again
   ↓
Continue Reasoning
   ↓
Final Answer

This is essential for complex questions that cannot be answered from a single piece of text.

Example
Question:
“Which products were developed by the team that the current CTO previously led?”
Multi-step reasoning process:

Step 1: Who is the current CTO?
→ Retrieved: “Bob is the current CTO.”
Step 2: Which team did Bob previously lead?
→ Retrieved: “Bob previously led Team Phoenix.”
Step 3: Which products did Team Phoenix develop?
→ Retrieved: “Team Phoenix developed Product X and Product Y.”
Final Answer: Product X and Product Y

This is multi-step reasoning over retrieved evidence.

Common Patterns
1. Retrieve → Reason → Retrieve → Reason
Most common pattern.
2. Plan → Execute Steps → Synthesize
Agent first creates a plan, then executes each step.
3. Evidence Collection → Intermediate Conclusions → Final Answer
Collect facts first, then reason over them.
