An AI agent  is a system that uses a large language model (LLM) as its brain to decide what to do next — instead of following a fixed, pre-programmed sequence of steps.

Normal chatbot / LLM prompt → always does roughly the same thing: takes your text → generates answer → done.The flow is always linear like give input prompt to LLM and llm provide  output and its done.
No thinking loop, no tool calls, no Api/DB call, no "let me check something first".

Agent → takes your goal → thinks → decides whether it needs more information / calculation / search/tool call  / code execution / email / database query / API call → does it → observes the result → thinks again → repeats until it believes the task is solved → gives final answer.

LangChain is an open-source framework for building agents. It lets you connect LLMs (like Claude, GPT, Gemini, etc.) to you API/Database, data sources, memory, and external systems so the model can read, take actions, and handle complex tasks instead of just chatting using your local system informations.

OutputParser is the component that transforms the messy/raw LLM output into the exact structured thing your code needs.

```python
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser 

model = init_chat_model("llama3.2", model_provider="ollama")

# Raw model call → returns AIMessage
raw_response = model.invoke("What is the capital of india?")

# Manual parse
parser = StrOutputParser()
clean_text = parser.invoke(raw_response)          # or parser.invoke(raw_response)
print(clean_text)
# Output example: "Golgappe and chole bhature are top choices in Delhi!"
```

Type of Parser

StrOutputParser → "just give me the words"

JsonOutputParser → "give me a dict from JSON"

PydanticOutputParser → "give me a real Python object with types & validation"

.with_structured_output() → "the modern clean way to get structured data"

```python
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import JsonOutputParser

model = init_chat_model("llama3.2", model_provider="ollama")

# Ask model to return JSON
prompt = "Return the answer in JSON format with key 'capital'. Question: What is the capital of India?"

raw_response = model.invoke(prompt)

# Extract text content
text = raw_response.content

# Parse JSON
parser = JsonOutputParser()
parsed = parser.invoke(text)

print(parsed)
```

PromptTemplate: For simple string-based prompts (best for older completion-style models like text-davinci).

ChatPromptTemplate: For modern chat/conversational models (recommended in most cases today). It uses a list of messages with roles like SystemMessage, HumanMessage, or AIMessage.

LCEL stands for LangChain Expression Language:-

LCEL  used the pipe operator  to connect prompt + model + parser  + … into one clean, reusable, debuggable expression.

A chain is a reusable recipe / pipeline that connects several steps together so that The output of step 1 automatically becomes the input of step 2 and 
The output of step 2 goes to step 3
… and so on. At the end you get exactly the result you want (clean text, JSON, structured object, etc.)

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.chat_models import init_chat_model

llm = init_chat_model("llama3.2", model_provider="ollama")
# 1. Create Prompt Template
prompt = ChatPromptTemplate.from_template(
    """You are a helpful assistant.
    Answer the question in a fun and engaging way.

    Question: {question}
    """
)


# 3. Create Output Parser
output_parser = StrOutputParser()

# 4. Build the Chain using LCEL (pipe | operator)
chain = prompt | llm | output_parser

# 5. Run the chain
response = chain.invoke({"question": "What is LangChain?"})

print(response)
```

Chain With stream processing

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.chat_models import init_chat_model

llm = init_chat_model("llama3.2", model_provider="ollama")
# 1. Create Prompt Template
prompt = ChatPromptTemplate.from_template(
    """You are a helpful assistant.
    Answer the question in a fun and engaging way.

    Question: {question}
    """
)
# 3. Create Output Parser
output_parser = StrOutputParser()

# 4. Build the Chain using LCEL (pipe | operator)
chain = prompt | llm | output_parser


# 3. Stream response (real-time)
for chunk in chain.stream({"question": "Explain quantum computing"}):
    print(chunk, end="", flush=True)
```

Chain with multiple input at one time

invoke() used for single input but batch() used for multiple input

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template("Tell me a short funny joke about {topic}")
llm = init_chat_model("llama3.2", model_provider="ollama")
parser = StrOutputParser()

chain = prompt | llm | parser

# Using batch - Pass list of inputs
topics = [
    {"topic": "Python"},
    {"topic": "AI"},
    {"topic": "LangChain"}
]

responses = chain.batch(topics)

for i, joke in enumerate(responses, 1):
    print(f"{i}. {joke}\n")
```

You can also control concurrency.Limit maximum number of parallel requests (useful for rate limits)

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template("Tell me a short funny joke about {topic}")
llm = init_chat_model("llama3.2", model_provider="ollama")
parser = StrOutputParser()

chain = prompt | llm | parser

# Using batch - Pass list of inputs
topics = [
    {"topic": "Python"},
    {"topic": "AI"},
    {"topic": "LangChain"}
]

responses = chain.batch(# Limit maximum number of parallel requests (useful for rate limits)
    inputs=topics,
    config={"max_concurrency": 3}   # Only 5 requests at a time
)

for i, joke in enumerate(responses, 1):
    print(f"{i}. {joke}\n")
```

In LangChain, context means the extra information you give to the LLM so it can answer accurately.
Think of it like this:

Without context → LLM answers from its own knowledge (can be wrong or outdated).You have to answer from memory only.

With context → You feed relevant information to the LLM, then ask the question.With context Teacher gives you notes + textbook pages (context), then asks you a question. Now you can give a much better answer.


```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.chat_models import init_chat_model

llm = init_chat_model("llama3.2", model_provider="ollama")
# 1. Create Prompt Template
prompt = ChatPromptTemplate.from_template(
    """Answer the question using the information given below.

Context: {context}          # ← This is where we put extra information

Question: {question}
"""
)
# 3. Create Output Parser
output_parser = StrOutputParser()

# 4. Build the Chain using LCEL (pipe | operator)
chain = prompt | llm | output_parser


response = chain.invoke({
    "context": "LangChain is a framework to build applications using LLMs. It was created in 2022.",
    "question": "What is LangChain?"
})
print(response)
```

You have two things that need to go to the LLM:

The question asked by the user
The context (some extra information)

But in a chain, data flows from left to right.
So how do you send both at the same time?
That’s why we use RunnablePassthrough().

```python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.chat_models import init_chat_model

# 1. Create the LLM
llm = init_chat_model("llama3.2", model_provider="ollama")

# 2. Create the prompt template
prompt = ChatPromptTemplate.from_template(
    """Answer the question using the following context:

Context: {context}

Question: {question}

Answer:"""
)

# 3. Create output parser
output_parser = StrOutputParser()

# 4. Build the chain
chain = (
    {
        "context": lambda x: "LangChain is a framework for building LLM apps. It makes chaining easy.",
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
    | output_parser
)

# 5. Run the chain
response = chain.invoke("What is LangChain?")
print(response)
```

Why do we write lambda x: ... in LCEL chains?

The dictionary { ... } must be a runnable — something that can accept input and produce output. If you write a plain string there like "context": "LangChain is a framework...", LangChain thinks:
"This is not a runnable → I cannot send input into it → error"
The dictionary must contain only runnables (or things that behave like runnables).
